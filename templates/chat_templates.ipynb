## Anthropic

```python
import anthropic

query = "Hello how are you! Could you help me with solving 3x + 2 = 62"

client = anthropic.Anthropic()

messages = [{'role': 'user', 'content': query}]

response = client.messages.create(max_tokens = 2024,
                                model = 'claude-3-7-sonnet-20250219', 
                                messages = messages)
```
